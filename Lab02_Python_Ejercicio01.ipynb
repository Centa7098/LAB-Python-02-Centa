```python
def multiply_3(x):
  num = lambda x : x % 3

  if num(x) == 0:
    return True
  else:
    return False

print(multiply_3(5))
print(multiply_3(3))
print(multiply_3(12))
```

```python
multiply_3 = lambda x: x % 3 == 0

print(multiply_3(5))
print(multiply_3(3))
print(multiply_3(12))
```

```python
cubo = lambda x : x ** 3

print(cubo(2))
print(cubo(3))
print(cubo(1))
```

```python
producto = lambda a,b : a*b

print(producto(2,3))
print(producto(2,0))
print(producto(5,4))
```

```python
greater = lambda a, b: a if a > b else b

print(greater(5, 9))
print(greater(20, 3))
```

```python
#Crear una lambda que reciba una palabra y devuelva True si empieza con la letra “A" (o “a”).

word = lambda text : True if text[0] == 'A' else False

print(word('Animal'))
print(word('Coconout'))
```

```python
#f. Crear una lambda que reciba una temperatura en Celsius y la convierta a Fahrenheit

Cel_Fah = lambda x : x*1.8 + 32

print(Cel_Fah(21))
print(Cel_Fah(0))
print(Cel_Fah(-32))
print(f"{Cel_Fah(21):.2f}")
```

```python
functions = [
    lambda x: x * 2,
    lambda x: x + 10,
    lambda x: x ** 2
]

num = int(input("Ingresa un número: "))

print(functions[0](num))
print(functions[1](num))
print(functions[2](num))
```
