```python
# a. Filtrar números pares y luego elevarlos al cuadrado.

def pares(num):
  return num

lista = [2, 6, 9, 7, 5, 4, 0, 3]
resultado = sorted(lista, key=pares)

new = []
for r in resultado:
  valor = r**2
  new.append(valor)

print(new)
```

```python
# b. Elevar cada número al cubo,filtrar los mayores que 100 y convertirlos a cadenas.

def cubo(num):
    return num**3

lista = [2, 1, 6, 5, 4, 3, 7, 9, 8]
ordenados = sorted(lista, key=cubo)

resultado = []
for num in ordenados:

    valor = num**3

    if valor > 100:
        resultado.append(str(valor))

print(resultado)
```

```python
#c. Dada una lista de numeros, filtrar números mayores que 10, elevarlos al cuadrado, ordenarlos de mayor a menor.

lista = [5,3,8,11,21,6,2,15,0,23,24,1]
higher_10 = [x**2 for x in lista if x > 10]
print(sorted(higher_10, reverse = True))
```
