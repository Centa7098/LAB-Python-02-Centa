```python
#a. Generar una lista con tuplas (n, n^2) para una lista de numeros.

lista = [2,4,6,7]
tuplas = [(n,n**2) for n in lista]

print(tuplas)
```

```python
#b. Generar una lista con cubos de los números mayores a 10.

lista = [2,5,10,12,21]
higher = [n for n in lista if n > 10]
print(higher)
```

```python
#c. Generar una lista con solo los numeros pares.

lista = [1,2,3,4,5,6,7,8,9,0]
odd_num = [n for n in lista if n % 2 == 0]
print(odd_num)
```

```python
#d. Dada la lista numeros, producir: ["valor: 10", "valor: 15", ...]

lista = [10,15,20,25,30]
text = [(f"valor: {str(n)}") for n in lista]
print(text)
```

```python
#e. Generar una lista con la conversion correspondiente a grados Fahrenheit.

lista = [-11,0,29]
fah = [n*1.8 + 32 for n in lista ]
print(fah)
```

```python
#f. Generar una lista filtrando las palabras de mas de 5 letras.

lista = ['Pizza','Hi','Bear','Chocolate','Infinite']
more_5 = [x for x in lista if len(x)>5]
print(more_5)
```

```python
#g. Generar una lista con solo las iniciales de las palabras.

lista = ['Pizza','Hi','Bear','Chocolate','Infinite']
init = [word[0] for word in lista]
for r in init:
  print(r)
```

```python
numbers = [1, 2, 3, 4, 5, 6]

result = ["par" if x % 2 == 0 else "impar" for x in numbers]

print(result)
```

```python
#i. Dada una lista de listas (matriz), aplanala. [[1,2,3], [4,5,6], [7,8,9]] = [1,2,3,4,5,6,7,8,9]

lista = [[1,2,3], [4,5,6], [7,8,9]]
new_matriz = [num for x in lista for num in x]
print(new_matriz)

no_matriz = [x for x in lista for num in x]
print(no_matriz)
```
